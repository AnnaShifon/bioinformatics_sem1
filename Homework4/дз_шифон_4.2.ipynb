# Задание 2. Сканирование хромосомы человека

```python
from Bio import motifs
from Bio import SeqIO
from Bio.Seq import Seq
```

```python
sites = ["GAGGTAAAC", "TCCGTAAGC", "CAGGTTGGA", "ACAGTCAGC", "TAGGTCAGC", 
         "CAGGTCAGC", "CAGGTCGAT", "CAGGTCAGC", "CAGGTCAGC", "CAGGTTGGC"]

instances = [Seq(site) for site in sites]
m = motifs.create(instances)
```

```python
pwm = m.counts.normalize(pseudocounts=0.8).log_odds()  #0.8 - это альфа, которую прибавляем что избежать неограниченного значения в матрице - примерно sqrt(N)/4 где N = 10
```

```python
records = SeqIO.parse("chr1.fa", "fasta") 
record = next(records)
sequence_all = record.seq
sequence = sequence_all[:1000000]
```

```python
hits = []

for position, score in pwm.search(sequence, threshold=5.0):
    if position >= 0:
        # Прямая цепь
        strand = "прямая"
        motif_seq = sequence[position:position + len(m)]
        hits.append((position, strand, score, str(motif_seq)))
    else:
        # Обратная цепь — Biopython возвращает отрицательную позицию
        strand = "обратная"
        # Истинная позиция на прямой цепи (левый конец)
        true_position = len(sequence) + position
        motif_seq = sequence[true_position:true_position + len(m)].reverse_complement()
        hits.append((true_position, strand, score, str(motif_seq)))
```

```python
print(f"{'Позиция':<10} {'Цепь':<8} {'Скор':<8} {'Последовательность'}")
for pos, strand, score, seq in hits:
    print(f"{pos:<10} {strand:<6} {score:<8.2f} {seq}")
```
